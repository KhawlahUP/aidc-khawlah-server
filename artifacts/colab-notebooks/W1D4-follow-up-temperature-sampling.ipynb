```python
import time
import torch
from transformers import AutoModelForCausalLM, AutoTokenizer

MODEL = "ibm-granite/granite-4.1-3b"

SHARED_PROMPT = "In one sentence, what is a data centre for?"
YOUR_PROMPT = "Explain what a GPU does, to someone who has never used one."

tokenizer = AutoTokenizer.from_pretrained(
    MODEL,
    trust_remote_code=True
)

model = AutoModelForCausalLM.from_pretrained(
    MODEL,
    torch_dtype=torch.float16,
    device_map="cuda",
    trust_remote_code=True
)

def ask(prompt, max_new_tokens=120):
    inputs = tokenizer(prompt, return_tensors="pt").to("cuda")

    start = time.time()

    outputs = model.generate(
        **inputs,
        max_new_tokens=max_new_tokens,
        do_sample=False
    )

    elapsed = time.time() - start

    answer = tokenizer.decode(
        outputs[0][inputs["input_ids"].shape[-1]:],
        skip_special_tokens=True
    ).strip()

    tokens = outputs.shape[-1] - inputs["input_ids"].shape[-1]

    print(f"\n--- {prompt}")
    print(answer)
    print(f"({tokens} tokens in {elapsed:.1f}s, {tokens / elapsed:.1f} tok/s)")

    return answer

print("model:", MODEL)

ask(SHARED_PROMPT)
ask(YOUR_PROMPT)
```

```python
def ask_sampled(prompt, max_new_tokens=120, temperature=0.8):
    inputs = tokenizer(prompt, return_tensors="pt").to("cuda")
    start = time.time()
    outputs = model.generate(
        **inputs,
        max_new_tokens=max_new_tokens,
        do_sample=True,
        temperature=temperature
    )
    elapsed = time.time() - start
    answer = tokenizer.decode(
        outputs[0][inputs["input_ids"].shape[-1]:],
        skip_special_tokens=True
    ).strip()
    tokens = outputs.shape[-1] - inputs["input_ids"].shape[-1]
    print(f"\n--- {prompt}")
    print(answer)
    print(f"({tokens} tokens in {elapsed:.1f}s, {tokens / elapsed:.1f} tok/s)")
    return answer
```

```python
prompts = [
    "What is a data centre for?",
    "What is a GPU for?",
    "What is a container for?",
]

# واحد تلو الثاني
t0 = time.time()
for p in prompts:
    ask(p, max_new_tokens=60)
print(f"sequential: {time.time()-t0:.1f}s")

# الكل دفعة وحدة
tokenizer.padding_side = "left"
if tokenizer.pad_token is None:
    tokenizer.pad_token = tokenizer.eos_token
t0 = time.time()
batch = tokenizer(prompts, return_tensors="pt", padding=True).to("cuda")
out = model.generate(**batch, max_new_tokens=60, do_sample=False)
print(f"batched:    {time.time()-t0:.1f}s")
for row in out:
    print(tokenizer.decode(row, skip_special_tokens=True)[:120])
```
